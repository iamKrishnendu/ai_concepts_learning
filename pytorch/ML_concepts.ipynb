# Linear Regression Class

```python
import torch
from torch.nn import Linear

torch.manual_seed(1) # it will give Random values for the Linear Class
model = Linear(in_features=1, out_features=1) # Linear Model : requires one Input_Size and one Output_Size
print(model.bias, model.weight)

x = torch.tensor([[2.0], [3.3]]) # Input to model (predictions)
print(model(x))
```

### Linear Regression: Custom Model

```python
import torch
import torch.nn as nn

class LR(nn.Module): # nn.Module : Inheritence a Parent class
    def __init__(self, input_size, output_size):
        super().__init__()
        self.linear = nn.Linear(input_size, output_size)
    
    def forward(self, x):
        pred = self.linear(x)
        return pred


torch.manual_seed(1) # bias
model = LR(1, 1) # using the Custom Linear Model
#print(list(model.parameters()))

x = torch.tensor([[1.0], [2.0]])
print(model.forward(x))
```

### Linear Regression: Creating Dataset

```python
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
import numpy as np

x = torch.randn(100, 1)*10 # returns a Tensor with random values
y = x + 3*torch.randn(100, 1)

plt.plot(x.numpy(), y.numpy(), 'o')
plt.ylabel('Y')
plt.xlabel('X')
#print(x)
```

```python
[w, b] = model.parameters()
print(w, b)

w1 = w[0][0].item()
b1 = b[0].item()
print(w1, b1)

def get_params():
    return (w[0][0].item(), b[0].item())

def plot_fit(title):
    plt.title = title
    w1, b1 = get_params()
    x1 = np.array([-30, 30])
    y1 = w1*x1 + b1
    plt.plot(x1, y1, 'r')
    plt.scatter(x, y)
    plt.show()

plot_fit('Initial Model')
```
